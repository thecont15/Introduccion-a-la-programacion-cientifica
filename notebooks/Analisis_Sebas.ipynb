# 4.3 Valores imposibles vs atípicos

```python
# Identificación de valores atípicos en edad mediante el rango intercuartílico
columna_analisis = "edad_"
q1, q3 = df_sivigila[columna_analisis].quantile([0.25, 0.75])
iqr = q3 - q1
limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

atipicos = df_sivigila[
    (df_sivigila[columna_analisis] < limite_inferior)
    | (df_sivigila[columna_analisis] > limite_superior)
]

print(f"Variable analizada: {columna_analisis}")
print(f"Primer cuartil (Q1): {q1:.2f}")
print(f"Tercer cuartil (Q3): {q3:.2f}")
print(f"IQR: {iqr:.2f}")
print(f"Límite inferior: {limite_inferior:.2f}")
print(f"Límite superior: {limite_superior:.2f}")
print(f"Valores atípicos encontrados: {len(atipicos)}")

atipicos[["id", "comuna", columna_analisis]].head()
```

 Como trabajare con los el dato de edad en cada comuna el valor atipico encontrado es 13 que es un valor bajo para el analisis

# 6.1 Imputacion por media/mediana/moda 

```python
# Imputación por media y mediana en todas las columnas numéricas
columnas_numericas = df_sivigila.select_dtypes(include="number").columns

for columna in columnas_numericas:
    media = df_sivigila[columna].mean()
    mediana = df_sivigila[columna].median()

    df_sivigila[f"{columna}_media"] = df_sivigila[columna].fillna(media)
    df_sivigila[f"{columna}_mediana"] = df_sivigila[columna].fillna(mediana)

    print(f"{columna}: media = {media:.2f}, mediana = {mediana:.2f}")

print(f"Columnas numéricas imputadas: {len(columnas_numericas)}")
df_sivigila[[f"{columna}_media" for columna in columnas_numericas]].head()
```

En mi caso no tuve que imputar datos ya que ninguna tupla estaba vacia

# 6.3 Imputacion que respeta la distribucion

# A. Muestreo de la distribución empírica (hot-deck aleatorio)

```python
import numpy as np

rng = np.random.default_rng(42)

for columna in columnas_numericas:
    observados = df_sivigila[columna].dropna()
    faltantes = df_sivigila[columna].isna()
    n_faltantes = faltantes.sum()

    if n_faltantes > 0 and not observados.empty:
        imputados = rng.choice(
            observados.to_numpy(),
            size=n_faltantes,
            replace=True,
        )
        df_sivigila.loc[faltantes, columna] = imputados

    print(f"{columna}: valores imputados aleatoriamente = {n_faltantes if not observados.empty else 0}")
```

# B. Media (o predicción) más ruido

```python
# Imputación de valores numéricos usando media y desviación estándar
rng = np.random.default_rng(42)

for columna in columnas_numericas:
    observados = df_sivigila[columna].dropna()
    faltantes = df_sivigila[columna].isna()
    n_faltantes = faltantes.sum()

    if n_faltantes > 0 and not observados.empty:
        media = observados.mean()
        desviacion = observados.std()
        desviacion = 0 if pd.isna(desviacion) else desviacion
        valores_imputados = rng.normal(
            media,
            desviacion,
            size=n_faltantes,
        )
        df_sivigila.loc[faltantes, columna] = valores_imputados

    print(f"{columna}: valores imputados con distribución normal = {n_faltantes if not observados.empty else 0}")
```

# 6.4 Imputación agrupada

```python
# Imputación agrupada por mediana en todas las columnas numéricas
columna_grupo = "comuna"

for columna in columnas_numericas:
    faltantes_antes = df_sivigila[columna].isna().sum()

    medianas_por_grupo = df_sivigila.groupby(columna_grupo)[columna].transform("median")
    df_sivigila[columna] = df_sivigila[columna].fillna(medianas_por_grupo)

    faltantes_despues = df_sivigila[columna].isna().sum()
    print(
        f"{columna}: faltantes antes = {faltantes_antes}, "
f"faltantes después = {faltantes_despues}"
)
```

En numero no cuento con ningun faltante

```python
import matplotlib.pyplot as plt
import numpy as np

# Preparar los datos de edad y ubicación
df_mapa = df_sivigila.copy()
df_mapa["edad_"] = pd.to_numeric(df_mapa["edad_"], errors="coerce")
df_mapa["comuna_mapa"] = (
    df_mapa["comuna"]
    .astype("string")
    .str.strip()
    .str.title()
    .fillna("Sin dato")
)

# Rangos de edad para analizar la edad de las madres
limites_edad = [0, 15, 20, 25, 30, 35, 40, 45, 100]
etiquetas_edad = [
    "0-14",
    "15-19",
    "20-24",
    "25-29",
    "30-34",
    "35-39",
    "40-44",
    "45+",
]
df_mapa["rango_edad"] = pd.cut(
    df_mapa["edad_"],
    bins=limites_edad,
    labels=etiquetas_edad,
    right=False,
)

# Frecuencia de madres por comuna y rango de edad
tabla_calor = pd.crosstab(
    df_mapa["comuna_mapa"],
    df_mapa["rango_edad"],
).reindex(columns=etiquetas_edad, fill_value=0)

fig, ax = plt.subplots(figsize=(12, 8))
imagen = ax.imshow(tabla_calor.values, cmap="YlOrRd", aspect="auto")

ax.set_title("Mapa de calor: edad de las madres por comuna de Medellín")
ax.set_xlabel("Rango de edad de la madre")
ax.set_ylabel("Comuna")
ax.set_xticks(np.arange(len(tabla_calor.columns)))
ax.set_xticklabels(tabla_calor.columns)
ax.set_yticks(np.arange(len(tabla_calor.index)))
ax.set_yticklabels(tabla_calor.index)

# Mostrar la cantidad de registros dentro de cada celda
for fila in range(tabla_calor.shape[0]):
    for columna in range(tabla_calor.shape[1]):
        valor = tabla_calor.iloc[fila, columna]
        ax.text(
            columna,
            fila,
            str(valor),
            ha="center",
            va="center",
            color="black" if valor < tabla_calor.values.max() / 2 else "white",
        )

fig.colorbar(imagen, ax=ax, label="Cantidad de registros")
plt.setp(ax.get_xticklabels(), rotation=45, ha="right")
fig.tight_layout()
plt.show()

tabla_calor
```

Al observar la grafica las madres que se encuentran en los barrios manrique, popular y villahermosa cuentan con el mayor flujo de registros con una edad de 20-24 años, los valores atipicos se encuentran entre menos de 14 años y mayores de 40 años

```python
import matplotlib.pyplot as plt

df_cajas = df_sivigila.copy()
df_cajas["talla_nacer"] = pd.to_numeric(
    df_cajas["talla_nacer"],
    errors="coerce",
)
df_cajas = df_cajas.dropna(subset=["talla_nacer"])

fig, axes = plt.subplots(1, 2, figsize=(13, 6))

# Distribución general de la talla al nacer
axes[0].boxplot(
    df_cajas["talla_nacer"],
    patch_artist=True,
    boxprops={"facecolor": "#9ecae1"},
    medianprops={"color": "#d62728", "linewidth": 2},
)
axes[0].set_title("Talla de los recién nacidos")
axes[0].set_ylabel("Talla al nacer (cm)")
axes[0].set_xticklabels(["Todos"])
axes[0].grid(axis="y", alpha=0.3)

# Distribución de la talla según el sexo registrado
grupos_sexo = []
etiquetas_sexo = []
for sexo, grupo in df_cajas.groupby("sexo_", dropna=False):
    valores = grupo["talla_nacer"].dropna()
    if not valores.empty:
        grupos_sexo.append(valores)
        etiquetas_sexo.append(str(sexo) if pd.notna(sexo) else "Sin dato")

axes[1].boxplot(
    grupos_sexo,
    patch_artist=True,
    boxprops={"facecolor": "#a1d99b"},
    medianprops={"color": "#d62728", "linewidth": 2},
)
axes[1].set_title("Talla al nacer por sexo")
axes[1].set_xlabel("Sexo registrado")
axes[1].set_ylabel("Talla al nacer (cm)")
axes[1].set_xticklabels(etiquetas_sexo)
axes[1].grid(axis="y", alpha=0.3)

fig.tight_layout()
plt.show()

print(f"Registros utilizados: {len(df_cajas)}")
print(f"Mediana de talla: {df_cajas['talla_nacer'].median():.2f} cm")
```

En la grafica de caja y bigotes podemos visualizar mejor las tallas al nacer con sus valores atipicos hay un estandar entre 46 y 47, entre los sexos femenino y masculino, y una dispersion muy grande, en cambio las tallas de las bebes femeninas se encuentran dispersos a cambio de las tallas del sexo masculino
