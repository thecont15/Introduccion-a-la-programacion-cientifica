# 4.3 Valores imposibles vs atípicos

```python
# Identificación de valores atípicos en edad mediante el rango intercuartílico
columna_analisis = "edad_"
q1, q3 = df_sivigila[columna_analisis].quantile([0.25, 0.75])
iqr = q3 - q1
limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

atipicos = df_sivigila[
    (df_sivigila[columna_analisis] < limite_inferior)
    | (df_sivigila[columna_analisis] > limite_superior)
]

print(f"Variable analizada: {columna_analisis}")
print(f"Primer cuartil (Q1): {q1:.2f}")
print(f"Tercer cuartil (Q3): {q3:.2f}")
print(f"IQR: {iqr:.2f}")
print(f"Límite inferior: {limite_inferior:.2f}")
print(f"Límite superior: {limite_superior:.2f}")
print(f"Valores atípicos encontrados: {len(atipicos)}")

atipicos[["id", "comuna", columna_analisis]].head()
```

# 6.1 Imputacion por media/mediana/moda 

```python
# Imputación por media y mediana en todas las columnas numéricas
columnas_numericas = df_sivigila.select_dtypes(include="number").columns

for columna in columnas_numericas:
    media = df_sivigila[columna].mean()
    mediana = df_sivigila[columna].median()

    df_sivigila[f"{columna}_media"] = df_sivigila[columna].fillna(media)
    df_sivigila[f"{columna}_mediana"] = df_sivigila[columna].fillna(mediana)

    print(f"{columna}: media = {media:.2f}, mediana = {mediana:.2f}")

print(f"Columnas numéricas imputadas: {len(columnas_numericas)}")
df_sivigila[[f"{columna}_media" for columna in columnas_numericas]].head()
```

# 6.3 Imputacion que respeta la distribucion

# A. Muestreo de la distribución empírica (hot-deck aleatorio)

```python
import numpy as np

rng = np.random.default_rng(42)

for columna in columnas_numericas:
    observados = df_sivigila[columna].dropna()
    faltantes = df_sivigila[columna].isna()
    n_faltantes = faltantes.sum()

    if n_faltantes > 0 and not observados.empty:
        imputados = rng.choice(
            observados.to_numpy(),
            size=n_faltantes,
            replace=True,
        )
        df_sivigila.loc[faltantes, columna] = imputados

    print(f"{columna}: valores imputados aleatoriamente = {n_faltantes if not observados.empty else 0}")
```

# B. Media (o predicción) más ruido

```python
# Imputación de valores numéricos usando media y desviación estándar
rng = np.random.default_rng(42)

for columna in columnas_numericas:
    observados = df_sivigila[columna].dropna()
    faltantes = df_sivigila[columna].isna()
    n_faltantes = faltantes.sum()

    if n_faltantes > 0 and not observados.empty:
        media = observados.mean()
        desviacion = observados.std()
        desviacion = 0 if pd.isna(desviacion) else desviacion
        valores_imputados = rng.normal(
            media,
            desviacion,
            size=n_faltantes,
        )
        df_sivigila.loc[faltantes, columna] = valores_imputados

    print(f"{columna}: valores imputados con distribución normal = {n_faltantes if not observados.empty else 0}")
```

# 6.4 Imputación agrupada

```python
# Imputación agrupada por mediana en todas las columnas numéricas
columna_grupo = "comuna"

for columna in columnas_numericas:
    faltantes_antes = df_sivigila[columna].isna().sum()

    medianas_por_grupo = df_sivigila.groupby(columna_grupo)[columna].transform("median")
    df_sivigila[columna] = df_sivigila[columna].fillna(medianas_por_grupo)

    faltantes_despues = df_sivigila[columna].isna().sum()
    print(
        f"{columna}: faltantes antes = {faltantes_antes}, "
f"faltantes después = {faltantes_despues}"
)
```

## Variables Predictivas

```python
import pandas as pd


def crear_variable_prematuridad(df):
    resultado = df.copy()
    resultado["sem_gest"] = pd.to_numeric(
        resultado["sem_gest"],
        errors="coerce",
    )
    resultado["prematuro"] = (resultado["sem_gest"] < 37).astype(int)
    return resultado


def test_menor_de_37_semanas_es_prematuro():
    df = pd.DataFrame({"sem_gest": [36]})

    resultado = crear_variable_prematuridad(df)

    assert resultado.loc[0, "prematuro"] == 1


def test_37_semanas_no_es_prematuro():
    df = pd.DataFrame({"sem_gest": [37]})

    resultado = crear_variable_prematuridad(df)

    assert resultado.loc[0, "prematuro"] == 0
```

```python
import pandas as pd


def test_menor_de_37_semanas_es_prematuro():
    df = pd.DataFrame({"sem_gest": [24, 32, 36]})
    resultado = crear_variable_prematuridad(df)
    assert resultado["prematuro"].tolist() == [1, 1, 1]


def test_37_semanas_o_mas_no_es_prematuro():
    df = pd.DataFrame({"sem_gest": [37, 38, 40]})
    resultado = crear_variable_prematuridad(df)
    assert resultado["prematuro"].tolist() == [0, 0, 0]


def test_convierte_semanas_de_texto_a_numeros():
    df = pd.DataFrame({"sem_gest": ["35", "37"]})
    resultado = crear_variable_prematuridad(df)
    assert resultado["prematuro"].tolist() == [1, 0]


test_menor_de_37_semanas_es_prematuro()
test_37_semanas_o_mas_no_es_prematuro()
test_convierte_semanas_de_texto_a_numeros()

print("Todas las pruebas fueron exitosas")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, KFold, cross_val_score
from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.dummy import DummyRegressor

RANDOM_STATE = 42

df = pd.read_csv("../data/processed/depuracion_sebas.csv.")
print(f"Filas: {df.shape[0]}, Columnas: {df.shape[1]}")
df.head()
```

#2. Selección de variables predictoras

```python
df = pd.read_csv("../data/processed/depuracion_sebas.csv")
print(f"Filas: {df.shape[0]}, columnas: {df.shape[1]}")

COLUMNAS_NUMERICAS = ["semana", "year_", "edad_", "sem_gest"]
COLUMNAS_CATEGORICAS = ["sexo_", "comuna", "tipo_ss_", "pac_hos_", "niv_edu_ma"]
OBJETIVO = "peso_nacer"

columnas_modelo = COLUMNAS_NUMERICAS + COLUMNAS_CATEGORICAS + [OBJETIVO]

# Verificar que todas las columnas existan antes de seleccionarlas
faltantes_columnas = [col for col in columnas_modelo if col not in df.columns]
if faltantes_columnas:
    raise ValueError(f"Faltan columnas en el CSV: {faltantes_columnas}")

datos_modelo = df[columnas_modelo].copy()

# Asegurar que predictores numéricos y objetivo sean numéricos
for columna in COLUMNAS_NUMERICAS + [OBJETIVO]:
    datos_modelo[columna] = pd.to_numeric(datos_modelo[columna], errors="coerce")

faltantes_antes = len(datos_modelo)
datos_modelo = datos_modelo.dropna(subset=columnas_modelo)
filas_descartadas = faltantes_antes - len(datos_modelo)

print(f"Filas antes de excluir faltantes: {faltantes_antes}")
print(
    f"Filas después: {len(datos_modelo)} "
    f"({filas_descartadas} descartadas, "
    f"{filas_descartadas / faltantes_antes * 100:.2f}%)"
)

# Codificar categorías para que scikit-learn pueda usarlas
X = pd.get_dummies(
    datos_modelo[COLUMNAS_NUMERICAS + COLUMNAS_CATEGORICAS],
    columns=COLUMNAS_CATEGORICAS,
    dtype=int,
)
y = datos_modelo[OBJETIVO]

print(f"Variables predictoras luego de codificar categorías: {X.shape[1]}")
```

Predictores usados (9 variables originales):

Numéricos: semana, year_, edad_, sem_gest
Categóricos: sexo_, comuna, tipo_ss_, pac_hos_, niv_edu_ma
Las categorías se convierten en columnas indicadoras con pd.get_dummies, por lo que la matriz X tendrá más de 9 columnas.

Objetivo: peso_nacer. Se predice, pero no es un predictor.

Variables excluidas del conjunto de predictores:
id, uni_med_, nombre_barrio, tip_cas_, cod_ase_, fec_con_, ini_sin_, talla_nacer, evento, barrio y prematuro.

Variables numéricas y categóricas ya definidas en el bloque anterior

```python
columnas_predictoras = COLUMNAS_NUMERICAS + COLUMNAS_CATEGORICAS

X = pd.get_dummies(
    datos_modelo[columnas_predictoras],
    columns=COLUMNAS_CATEGORICAS,
    drop_first=True,
    dtype=int,
)

y = datos_modelo[OBJETIVO]

print(f"Predictores finales: {X.shape[1]} columnas tras codificar categorías")
print(f"Variable objetivo: {OBJETIVO}")
X.head()
```

3. Particion entrenamiento / prueba

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=RANDOM_STATE,
)

print(f"Entrenamiento: {X_train.shape[0]} filas")
print(f"Prueba: {X_test.shape[0]} filas")
print(f"Predictores por fila: {X_train.shape[1]}")
```

Separa el 80 % de los registros para entrenamiento y el 20 % para prueba.

4. Modelo de referencia (baseline)

```python
baseline = DummyRegressor(strategy="mean")
baseline.fit(X_train, y_train)

pred_baseline = baseline.predict(X_test)

mae_baseline = mean_absolute_error(y_test, pred_baseline)
rmse_baseline = np.sqrt(mean_squared_error(y_test, pred_baseline))

print(
    f"Baseline (predice el peso promedio) -> "
    f"MAE: {mae_baseline:.1f} g | RMSE: {rmse_baseline:.1f} g"
)
```

5. Selección de profundidad del árbol (validación cruzada)

```python
profundidades = list(range(2, 16))
cv = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

mae_por_profundidad = []

for profundidad in profundidades:
    modelo_cv = DecisionTreeRegressor(
        max_depth=profundidad,
        random_state=RANDOM_STATE,
    )
    puntajes = cross_val_score(
        modelo_cv,
        X_train,
        y_train,
        cv=cv,
        scoring="neg_mean_absolute_error",
    )
    mae_por_profundidad.append(-puntajes.mean())

mejor_profundidad = profundidades[int(np.argmin(mae_por_profundidad))]
print(f"Profundidad óptima según validación cruzada: {mejor_profundidad}")
print(f"MAE promedio: {min(mae_por_profundidad):.1f} g")

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(profundidades, mae_por_profundidad, marker="o", color="#2E5A87")
ax.axvline(
    mejor_profundidad,
    color="black",
    linestyle="--",
    linewidth=1,
    label=f"Óptimo = {mejor_profundidad}",
)
ax.set_title("MAE de validación cruzada según profundidad del árbol")
ax.set_xlabel("Profundidad máxima")
ax.set_ylabel("MAE (g)")
ax.legend()
plt.tight_layout()
plt.show()
```

6. Entrenamiento del modelo final

```python
modelo = DecisionTreeRegressor(
    max_depth=mejor_profundidad,
    random_state=RANDOM_STATE,
)

modelo.fit(X_train, y_train)

# Predicciones para evaluar desempeño en entrenamiento y prueba
pred_train = modelo.predict(X_train)
pred_test = modelo.predict(X_test)

print(f"Objetivo: {OBJETIVO} (peso al nacer, en gramos)")
print(f"Profundidad máxima elegida: {mejor_profundidad}")
print(f"Registros de entrenamiento: {len(pred_train)}")
print(f"Registros de prueba: {len(pred_test)}")
```

7. Evaluación del modelo

```python
def metricas(y_real, y_pred, nombre):
    mae = mean_absolute_error(y_real, y_pred)
    rmse = np.sqrt(mean_squared_error(y_real, y_pred))
    r2 = r2_score(y_real, y_pred)

    print(
        f"{nombre:12s} -> "
        f"MAE: {mae:6.1f} g | RMSE: {rmse:6.1f} g | R²: {r2:.3f}"
    )
    return mae, rmse, r2


mae_train, rmse_train, r2_train = metricas(
    y_train, pred_train, "Entrenamiento"
)
mae_test, rmse_test, r2_test = metricas(
    y_test, pred_test, "Prueba"
)

if mae_baseline != 0:
    mejora = (1 - mae_test / mae_baseline) * 100
    print(f"\nMejora frente al baseline en MAE: {mejora:.1f}%")
else:
    print("\nNo se puede calcular la mejora: el MAE del baseline es 0.")
```

```python
fig, ax = plt.subplots(figsize=(6, 6))

ax.scatter(
    y_test,
    pred_test,
    alpha=0.3,
    s=15,
    color="#B85C38",
)

limite = [
    min(y_test.min(), pred_test.min()),
    max(y_test.max(), pred_test.max()),
]
ax.plot(
    limite,
    limite,
    color="black",
    linestyle="--",
    linewidth=1,
    label="Predicción perfecta",
)

ax.set_title("Peso al nacer real vs. predicho (conjunto de prueba)")
ax.set_xlabel("Peso real (g)")
ax.set_ylabel("Peso predicho (g)")
ax.legend()
plt.tight_layout()
plt.show()
```

Segun la grafica indica una prediccion en el peso real y el predicho, entre 2200 y 2400, con un error medio de 3.9 con el baseline

8. Importancia de variables

```python
importancias = pd.Series(
    modelo.feature_importances_,
    index=X.columns,
)

importancias = (
    importancias[importancias > 0]
    .sort_values(ascending=True)
    .tail(12)
)

fig, ax = plt.subplots(figsize=(8, 5))
importancias.plot(kind="barh", ax=ax, color="#5B8C5A")
ax.set_title(f"Variables más importantes para predecir {OBJETIVO}")
ax.set_xlabel("Importancia (reducción de impureza)")
plt.tight_layout()
plt.show()
```

9. Visualización del árbol (versión interpretable, profundidad 3)

```python
modelo_interpretable = DecisionTreeRegressor(
    max_depth=3,
    random_state=RANDOM_STATE,
)
modelo_interpretable.fit(X_train, y_train)

fig, ax = plt.subplots(figsize=(18, 8))
plot_tree(
    modelo_interpretable,
    feature_names=X_train.columns,
    filled=True,
    rounded=True,
    fontsize=8,
    ax=ax,
)
ax.set_title("Árbol interpretable para predecir el peso al nacer")
plt.tight_layout()
plt.show()
```
