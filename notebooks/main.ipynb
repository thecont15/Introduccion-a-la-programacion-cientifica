# 2.2 LECTURA DE DATA CONTROLADA

```python

import pandas as pd

# Archivo descargado desde World Bank Data360
RUTA_WORLD_BANK = "../data/WB_CLEAR_AQUASTAT_4554.csv"

df_world_bank = pd.read_csv(
    RUTA_WORLD_BANK,
    sep=",",
    encoding="utf-8",
    decimal=".",
    na_values=["", "NA", "N/A", "-", "null", "999"],
    skipinitialspace=True,
)

print(f"Filas: {df_world_bank.shape[0]}")
print(f"Columnas: {df_world_bank.shape[1]}")
df_world_bank.head()
```

# 3.PARCEAR DATOS

# 3.1 NUMEROS

```python
# Conversión de las columnas numéricas del dataset

df_world_bank["TIME_PERIOD"] = pd.to_numeric(
    df_world_bank["TIME_PERIOD"],
    errors="coerce",
)
df_world_bank["OBS_VALUE"] = pd.to_numeric(
    df_world_bank["OBS_VALUE"],
    errors="coerce",
)

print(df_world_bank[["TIME_PERIOD", "OBS_VALUE"]].dtypes)
df_world_bank[["REF_AREA_LABEL", "TIME_PERIOD", "OBS_VALUE"]].head()
```

# NUMEROS PERDIDOS

```python
perdidos_en_parseo = df_world_bank["OBS_VALUE"].isna().sum()
perdidos_en_parseo_ = df_world_bank["TIME_PERIOD"].isna().sum()


print(f"Valores perdidos en OBS_VALUE: {perdidos_en_parseo}")
print(f"Valores perdidos en TIME_PERIOD: {perdidos_en_parseo_}")
```

# 3.2 TEXTO Y CATEGORIAS

```python
# Normalización de todas las columnas de texto
columnas_texto = df_world_bank.select_dtypes(include=["str"]).columns

for columna in columnas_texto:
    df_world_bank[columna] = (
        df_world_bank[columna]
        .astype("string")
        .str.strip()
        .str.lower()
        .replace("nan", pd.NA)
    )

print(f"Columnas de texto normalizadas: {len(columnas_texto)}")
df_world_bank[columnas_texto].head()
```

# PERDIDOS TEXTO Y CATEGORIAS

```python
# Valores perdidos en todas las columnas de texto
perdidos_texto = df_world_bank[columnas_texto].isna().sum()

print("Valores perdidos por columna de texto:")
print(perdidos_texto)
```

# 4.2 Duplicados

```python
# Eliminación de copias exactas del dataset World Bank
filas_antes = len(df_world_bank)
df_world_bank = df_world_bank.drop_duplicates(keep="first")
filas_despues = len(df_world_bank)

print(f"Filas antes de eliminar duplicados: {filas_antes}")
print(f"Duplicados eliminados: {filas_antes - filas_despues}")
print(f"Filas después de eliminar duplicados: {filas_despues}")
```

# 4.3 Valores imposibles vs atípicos

```python
# Identificación de valores atípicos en OBS_VALUE mediante el rango intercuartílico
q1, q3 = df_world_bank["OBS_VALUE"].quantile([0.25, 0.75])
iqr = q3 - q1
limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

atipicos = df_world_bank[
    (df_world_bank["OBS_VALUE"] < limite_inferior)
    | (df_world_bank["OBS_VALUE"] > limite_superior)
]

print(f"Primer cuartil (Q1): {q1:.2f}")
print(f"Tercer cuartil (Q3): {q3:.2f}")
print(f"IQR: {iqr:.2f}")
print(f"Límite inferior: {limite_inferior:.2f}")
print(f"Límite superior: {limite_superior:.2f}")
print(f"Valores atípicos encontrados: {len(atipicos)}")

atipicos[["REF_AREA_LABEL", "TIME_PERIOD", "OBS_VALUE"]].head()
```

# 5.2 Cuando si se puede eliminar datos

```python
# Filas sin ningún valor
df_world_bank = df_world_bank.dropna(how="all")

# Columna casi vacía: se documenta y se deja fuera del análisis
umbral = 0.40
columnas_utiles = df_world_bank.columns[df_world_bank.isna().mean() < umbral]
columnas_excluidas = df_world_bank.columns.difference(columnas_utiles)
df_world_bank = df_world_bank[columnas_utiles]

print(f"Columnas conservadas: {len(columnas_utiles)}")
print(f"Columnas excluidas: {len(columnas_excluidas)}")
```

# 6.1 Imputacion por media/mediana/moda 

```python
# Imputación de valores numéricos en OBS_VALUE
media = df_world_bank["OBS_VALUE"].mean()
df_world_bank["OBS_VALUE_media"] = df_world_bank["OBS_VALUE"].fillna(media)

mediana = df_world_bank["OBS_VALUE"].median()
df_world_bank["OBS_VALUE_mediana"] = df_world_bank["OBS_VALUE"].fillna(mediana)

# Imputación de valores categóricos en REF_AREA_LABEL
moda = df_world_bank["REF_AREA_LABEL"].mode(dropna=True)
df_world_bank["REF_AREA_LABEL_moda"] = df_world_bank["REF_AREA_LABEL"].fillna(
    moda.iloc[0] if not moda.empty else pd.NA
)

print(f"Media de OBS_VALUE: {media:.2f}")
print(f"Mediana de OBS_VALUE: {mediana:.2f}")
print(f"Moda de REF_AREA_LABEL: {moda.iloc[0] if not moda.empty else 'Sin moda'}")
```

# 6.3 Imputacion que respeta la distribucion

# A. Muestreo de la distribución empírica (hot-deck aleatorio)

```python
import numpy as np

obs = df_world_bank["OBS_VALUE"].dropna()
n_faltan = df_world_bank["OBS_VALUE"].isna().sum()

rng = np.random.default_rng(42)

if n_faltan > 0 and not obs.empty:
    imputados = rng.choice(obs.to_numpy(), size=n_faltan, replace=True)
    df_world_bank.loc[df_world_bank["OBS_VALUE"].isna(), "OBS_VALUE"] = imputados

print(f"Valores imputados aleatoriamente: {n_faltan if not obs.empty else 0}")
```

# B. Media (o predicción) más ruido

```python
obs = df_world_bank["OBS_VALUE"].dropna()
mu, sigma = obs.mean(), obs.std()
n_faltan = df_world_bank["OBS_VALUE"].isna().sum()

if n_faltan > 0 and not obs.empty:
    valores_imputados = rng.normal(mu, sigma, size=n_faltan)
    df_world_bank.loc[df_world_bank["OBS_VALUE"].isna(), "OBS_VALUE"] = valores_imputados

print(f"Valores imputados con distribución normal: {n_faltan if not obs.empty else 0}")
```

# 6.4 Imputación agrupada

```python
faltantes_antes = df_world_bank["OBS_VALUE"].isna().sum()

df_world_bank["OBS_VALUE"] = df_world_bank.groupby("REF_AREA_LABEL")["OBS_VALUE"].transform(
    lambda serie: serie.fillna(serie.median())
)

faltantes_despues = df_world_bank["OBS_VALUE"].isna().sum()
print(f"Valores faltantes antes: {faltantes_antes}")
print(f"Valores faltantes después: {faltantes_despues}")
```
